# Egenværdier og egenvektorer og løsning af differentialligningssystemer med SymPy
Denne notebook demonstrerer, hvordan man finder egenværdier, egenvektorer og løser differentialligninger med med SymPy.

```python
import sympy as sp

sp.init_printing()  # gør at matematiske udtryk skrives pænt

# Definer en matrix
A = sp.Matrix([[3, -2], [5, -3]])
A
```

## Beregning af egenværdier

```python
eigenvals = A.eigenvals()
eigenvals
```

## Beregning af egenvektorer

```python
eigenvects = A.eigenvects()
eigenvects
```

Hvad angiver mon de forskellige tal og vektorer i udtrykket ovenfor? Man kan i udtrykket både aflæse algebraiske og geometriske multipliciteter, find ud af hvordan.
Prøv nu at regne opgave 2, 4, 6 og 7  fra StoreDag uge 11 ved hjælp af ovenstående kommandoer, men prøv også via SymPy at finde og løse det karakteristiske polynomium for en af opgaverne, i kender allerede SymPy kommandoen til at finde determinanten med.
Sammenlign resultatet med den manuelle beregning.

```python
#OPG 2
A = sp.Matrix([
    [-5, 10],
    [1, 4]
])
I2 = sp.eye(2)
Z = sp.Symbol('Z')
pA = sp.det(A - Z*I2)
pA = pA.factor()
display(pA)
eigenvals = A.eigenvals()
eigenvals
```

```python
#OPG 4
A = sp.Matrix([
    [2, 2],
    [-1, 4]
])
# a)
I2 = sp.eye(2)
Z = sp.Symbol('Z')
pA = sp.det(A - Z*I2)
pA = pA.factor()
display(pA)
eigenvals = A.eigenvals()
display(eigenvals)

# b) 
i = sp.I
eigenspace_3i = (A - (3+i) * I2).nullspace()
eigenspace_3i
```

```python
A = sp.Matrix([
    [1, -1, 1],
    [2, 4, -1],
    [0, 0, 3]
])
# a)
Z = sp.Symbol('Z')
I3 = sp.eye(3)
pA = sp.det(A - Z*I3)
pA = pA.factor()
display(pA)

# b)
eigenspace_3 = (A - 3*I3).nullspace()
eigenspace_2 = (A - 2*I3).nullspace()
display(eigenspace_3, eigenspace_2)
```

```python
#OPG 7
B = sp.Matrix([
    [1, 1, 0],
    [2, -1, -1],
    [0, 2, 1]
])
pA = sp.det(B - Z*I3)

# b)
eigenvals = B.eigenvals()
display(eigenvals)

# c)
eigenspace_m1 = (B + I3).nullspace()
eigenspace_1 = (B - I3).nullspace()
display(eigenspace_m1, eigenspace_1)
```

# Systemer af differentialligninger i SymPy


```python
t = sp.symbols('t')
x1 = sp.Function('x1')
x2 = sp.Function('x2')
```

## System:

```python
eq1 = sp.Eq(sp.diff(x1(t), t), -x1(t) + 2*x2(t))
eq2 = sp.Eq(sp.diff(x2(t), t), -2*x2(t))
display(eq1)
display(eq2)
```

## Fuldstændig løsning

```python
sol = sp.dsolve((eq1, eq2))
display(sol)
```

## Løsning med begyndelsesværdier

```python
ics = {x1(0):3, x2(0):4}
sol_ics = sp.dsolve((eq1, eq2), ics=ics)
display(sol_ics)
```

```python
# OPG 1
t = sp.symbols('t', real=True)

# a)
f = t
eq1 = sp.Eq(sp.diff(f, t)**2 - 4*sp.exp(2*t)*f, 0)
display(eq1)

# b)
f = sp.exp(2*t)
eq2 = sp.Eq(sp.diff(f, t)**2 - 4*sp.exp(2*t)*f, 0)
display(eq2)
```

```python
# OPG2
a, b, t = sp.symbols('a b t')
f_p = a*t + b

display(sp.Eq(sp.diff(f_p, t) - 3*f_p, t).expand())

eq1 = sp.Eq(-3*a, 1)
eq2 = sp.Eq(a - 3*b, 0)

sp.solve((eq1, eq2), (a, b))
```
