```python
import sympy as sp
```

```python
A = sp.Matrix([
    [1, 2, 14, 7], 
    [2, 2, 20, 16], 
    [-1, -1, -10, -8], 
    [1, -1, 2, 10]
    ])
n = 4
dim_ker = n - A.rank()
print('Dimension af kerne:', dim_ker)
```

```python
x1,x2,x3 = sp.symbols('x1,x2,x3')
a,b = sp.symbols('a,b')

eq1 = sp.Eq(x1 + 2 * x3, b)
eq2 = sp.Eq((-1 + a) * x2 - 2 * x3, -b)
eq3 = sp.Eq(a * x1 + a * x3, a)

A = sp.Matrix([
    [1, 0, 2],
    [0, -1, -2],
    [0, 0, 0]
])
b = sp.Matrix([2, -2, 0])
T = A.row_join(b)
print('Rang af A:', A.rank())
print('Rang af T:', T.rank())
```

```python
sol = sp.linsolve((A,b))
display(sol)
```
