## 2. linsolve – Klarer både ligningerne direkte eller omskrevet på matrix form 

`linsolve()` bruges til at finde løsningen til lineære ligningssystemer.

```python
# Definer variabler
x, y, z = sp.symbols('x y z')

# Definer ligningerne
eq1 = x + 2*y + z - 4
eq2 = 2*x + y - z - 1

# Løs systemet
solution = sp.linsolve([eq1, eq2], (x, y, z))
solution
```

```python
# Definer matricer
A = sp.Matrix([
    [1, 2, 1],
    [2, 1, -1]
])
b = sp.Matrix([4, 1])
display(A, b)
# Løs systemet Ax = b
solution = sp.linsolve((A, b))
solution
```

## 3. Rang og Lineær uafhængighed

En samling af vektorer er *lineært uafhængige*, hvis ingen af dem kan skrives som en linearkombination af de andre.

Vi kan teste dette ved hjælp af kommandoen `sp.rank()` som finder rangen af en matrix.

```python
v1 = sp.Matrix([1, 2, 3]) # vektor 1
v2 = sp.Matrix([2, 4, 6]) # vektor 2
v3 = sp.Matrix([1, 0, -1]) # vektor 3

M = sp.Matrix.hstack(v1, v2, v3) # Samler de tre vektorer i en matrix
print("Matrix af vektorer:")
display(M)
print("Rang af M:", M.rank()) # Finder rangen
```

**Fortolkning:**
- Hvis rang = antal vektorer → lineært uafhængige.
- Hvis rang < antal vektorer → afhængige.

## 4. Matrixalgebra

Vi kan udføre almindelige matrixoperationer som addition, multiplikation og transponering.

```python
A = sp.Matrix([[1, 2], [3, 4]])
B = sp.Matrix([[2, 0], [1, 2]])

print("A:")
display(A)
print("B:")
display(B)

print("A + B =")
display(A + B) 

print("A * B =")
display(A * B)

print("A transponeret:")
display(A.T)
```

**Øvelse 4:**  
Prøv nu at regne opgave 4 StoreDag og opgave 1 LilleDag uge 7 i SymPy, med brug af ovenstående kommandoer, og sammenlign resultatet med den manuelle beregning.

```python
#OPG4

A = sp.Matrix([
    [1,0,0],
    [0,1,0]
])
B = sp.Matrix([
    [1,4],
    [2,1],
    [3,0]
])
C = sp.Matrix([
    [1,0,0],
    [0,-1,0],
    [2,0,0]
])
display(A*B,B*A)
display((-1/2)*C)
display((-1/2)*C+B*A)
```

```python
# OPG1
A = sp.Matrix([
    [1,2],
    [3,4],
    [5,6]
])
display(A,A.T)
display(A.T*A)
```

## 5. Invers matrix

Hvis en matrix har en **invers**, kan den inverse findes i SymPy med `.inv()`.

```python
A = sp.Matrix([[1, 2], [3, 4]])
A_inv = A.inv()

print("A:")
display(A)
print("Invers af A:")
display(A_inv)

print("A * A_inv =")
display(sp.simplify(A * A_inv))
```
